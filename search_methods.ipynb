```python
from typing import List

class Solution:
    def search(self, nums: List[int], target: int) -> int:
        left = 0
        right = len(nums) - 1
        while left <= right:
            mid = (right - left) // 2
            if nums[mid] == target:
                return mid
            if nums[mid] > nums[right]:
                left = mid + 1
            else:
                right = mid - 1
        
        return -1
    
#Test
solution = Solution()
print(solution.search([4,5,6,7,0,1,2], 0))  # Output: 4
```

```python
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left = 0
        right = len(nums) - 1 
        res = []

        while left <= right:
            mid = (left + right) // 2
            
            if nums[mid] == target:
                res.append(mid)
                left = mid + 1
            
            if nums[mid] < target:
                left = mid + 1
            else:
                right = mid - 1

        res.sort()
        return res

solution = Solution()
print(solution.searchRange([5,7,7,8,8,8,10], 8))  # Output: [3, 4]
```

```python
class Solution:
    def isValidSudoku(self, board: List[List[str]]) -> bool:
        rows = {i: set() for i in range(9)}
        cols = {j: set() for j in range(9)}
        boxes = {(i, j): set() for i in range(3) for j in range(3)}
        for i in range(9):
            for j in range(9):
                if board[i][j] == ".":
                    continue
                    
                if board[i][j] in rows[i] or board[i][j] in cols[j] or board[i][j] in boxes[(i // 3, j //3)]:
                    return False
            
                rows[i].add(board[i][j])
                cols[j].add(board[i][j])
                boxes[(i // 3, j // 3)].add(board[i][j])
        return True
    
#Test
solution = Solution()
print(solution.isValidSudoku([
  ["5","3",".",".","7",".",".",".","."],
  ["6",".",".","1","9","5",".",".","."],
  [".","9","8",".",".",".",".","6","."],
  ["8",".",".",".","6",".",".",".","3"],
  ["4",".",".","8",".","3",".",".","1"],
  ["7",".",".",".","2",".",".",".","6"],
  [".","6",".",".",".",".","2","8","."],
  [".",".",".","4","1","9",".",".","5"],
  [".",".",".",".","8",".",".","7","9"]
]))  # Output: True
```
