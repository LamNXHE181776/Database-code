```python
import numpy as np
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def isEmpty(self):
        return self.size == 0
    
    def length(self):
        return self.size
    
    def enqueue(self, data):
        new_node = Node(data)
        if self.tail is None:
            self.head = self.tail = new_node
            self.size += 1
            return        

        self.tail.next = new_node
        self.tail = self.tail.next
        self.size += 1
        return
    
    def dequeue(self):
        if self.isEmpty():
            return "Empty"
        
        if self.size == 1:
            temp = self.head.data 
            self.head = self.tail = None
            self.size -= 1
            return temp
        
        temp = self.head
        self.head = self.head.next
        self.size -= 1
        return temp.data
    
    def peek(self):
        if self.isEmpty():
            return "Empty"
        
        return self.head.data
    
    def print_queue(self):
        temp = self.head
        while temp:
            print(temp.data, end='=>')
            temp = temp.next
        print()

    def find_lowest_value(self):
        if self.isEmpty():
            return "Empty"

        min_value = self.head.data
        current_node = self.head.next

        while current_node:
            if current_node.data < min_value:
                min_value = current_node.data
            current_node = current_node.next

        return min_value    
    
    def delete_value(self, value):
        if self.isEmpty():
            return "Empty"
        
        if value == self.head.data:
            deleted = self.head.data
            self.head = self.head.next
            self.size -= 1

            if self.head is None:
                self.tail = self.head

            return deleted
        
        current_node = self.head
        
        while current_node.next and current_node.next.data != value:
            current_node = current_node.next

        if current_node.next is None:
            return "Not Found"
        
        deleted = current_node.next.data
        if current_node.next == self.tail:
            self.tail = current_node

        current_node.next = current_node.next.next
        self.size -=1
        return deleted
    
    def insert(self, value, position):
            if position < 1 or position > self.size + 1:
                raise ValueError
            
            new_node = Node(value)
            current_node = self.head

            if position == 1:
                new_node.next = self.head
                self.head = new_node
                if self.size == 0:
                    self.tail = new_node
                self.size += 1
                return

            for _ in range(position - 2):
                if current_node is None:
                    return "Not found"
                current_node = current_node.next

            new_node.next = current_node.next
            current_node.next = new_node

            if new_node.next is None:
                self.tail = new_node

            self.size +=1
            return
```

```python
myQueue = LinkedList()

for _ in range(1, 20):
    myQueue.enqueue(np.random.randint(1,100))

print("Queue: ", end="")
myQueue.print_queue()
print("Peek: ", myQueue.peek())
print("Dequeue: ", myQueue.dequeue())
print("Queue after Dequeue: ", end="")
myQueue.print_queue()
print("isEmpty: ", myQueue.isEmpty())
print("Size: ", myQueue.length()) 
```
