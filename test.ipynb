```python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        m = {')':'(',']':'[','}':'{'}

        for c in s:
            if c in m:
                if not stack or stack.pop() != m[c]:
                    return False
            else:
                stack.append(c)
        return not stack
#test

s = "()[]{}"
sol = Solution()
print(sol.isValid(s))
```

```python
class Solution:
    def removeDuplicates(self, nums: list[int]) -> int:
        nums_unique = set(nums) 
        nums_unique = list(nums_unique)
        nums_unique = sorted(nums_unique)

        k = len(nums_unique)

        for i in range(len(nums) - len(nums_unique)):
            nums_unique += ['_']
        
        return k, nums_unique
```

```python
from typing import List
class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        if len(nums) == 1:
            return [nums[:]]
        
        res = []

        for _ in range(len(nums)):
            n = nums.pop(0)
            perms = self.permute(nums)

            for p in perms:
                p.append(n)
            
            res.extend(perms)
            nums.append(n)
        
        return res

sol = Solution()
print(sol.permute([1,2,3]))
```

```python
from collections import Counter
from typing import List

class Solution:
    def permuteUnique(self, nums: List[int]) -> List[List[int]]:
        res = []

        def backtrack(counter, comb):
            if len(comb) == len(nums):
                res.append(list(comb))
                return 

            for i in counter:
                if counter[i] > 0:
                    comb.append(i)
                    counter[i] -= 1

                    backtrack(counter, comb)

                    comb.pop()
                    counter[i] += 1
            

        backtrack(Counter(nums), [])

        return res
    
sol = Solution()
print(sol.permuteUnique([1,2,3]))
```

```python

from collections import Counter
from typing import List

class Solution:

    def rotate(self, matrix: List[List[int]]) -> None:
            """
            Do not return anything, modify matrix in-place instead.
            """
            for i in range(len(matrix)):
                for j in range(len(matrix[i])):
                    if i < j:
                        matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
                matrix[i].reverse()
            return matrix

sol = Solution()
print(sol.rotate([[1,2,3],[4,5,6],[7,8,9]]))
```

```python
class Solution:
    def maxSubArray(self, nums: list[int]) -> int:
        if len(nums) == 1:
            return nums[0]
        res = nums[0]
        max_end = nums[0]
        for i in range(1, len(nums)):
            max_end = max(nums[i], max_end + nums[i])
            
            res = max(res, max_end)

        return res
    
sol = Solution()
x = [-2, -1]
print(sol.maxSubArray(x))
```

```python
from typing import List

class Solution:
    def canJump(self, nums: List[int]) -> bool:
        n = len(nums)
        i = 0
        while i <= n - 1:     
            i += nums[i]    
            if i < n - 1:
                continue
            elif i > n - 1:
                return False
            else: return True

            

sol = Solution()
x = [2, 3, 1, 1, 4]
print(sol.canJump(x))
```

```python
class Solution:
    def canJump(self, nums: List[int]) -> bool:

        farthest = 0

        for i in range(len(nums)):
            if i > farthest:
                return False

            farthest = max(farthest, i + nums[i])

            if farthest >= len(nums) - 1:
                return True
        
        return True

    
sol = Solution()
x = [3,2,1,0,4]
print(sol.canJump(x))
```

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort()
        output = []
        output.append(intervals[0])

        for i in range(1, len(intervals)):
            last = output[-1]
            curr = intervals[i]

            if curr[0] <= last[-1]:
                if curr[-1] > last[-1]:
                    last[-1] = curr[-1]
            else:
                output.append(curr)

        return output
    
sol = Solution()
x = [[1,3],[2,6],[8,10],[15,18]]
print(sol.merge(x))
```

```python
#Binary search to return location of target in sorted array, return theoretical location if not found
def binary_search(nums, target):
    left = 0
    right = len(nums) - 1

    while left <= right:
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    
    return left
```
