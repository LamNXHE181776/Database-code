```python
import numpy as np
```

```python
#QUEUE

class Queue:
    def __init__(self):
        self.queue = []

    def isEmpty(self):
        return len(self.queue) == 0

    def enqueue(self, value):
        self.queue.append(value)
        return value
    
    def dequeue(self):
        if self.isEmpty():
            return "Queue is empty"
        value = self.queue[0]
        self.queue.pop(0)
        return value
    
    def peek(self):
        if self.isEmpty():
            return "Queue is empty"
        return self.queue[0]
    
    def size(self):
        return len(self.queue)
```

```python
 # Create a queue
myQueue = Queue()

for _ in range(1, 20):
    myQueue.enqueue(np.random.randint(1,100))

print("Queue: ", myQueue.queue)
print("Peek: ", myQueue.peek())
print("Dequeue: ", myQueue.dequeue())
print("Queue after Dequeue: ", myQueue.queue)
print("isEmpty: ", myQueue.isEmpty())
print("Size: ", myQueue.size()) 
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedQueue:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def isEmpty(self):
        return self.size == 0
    
    def length(self):
        return self.size
    
    def enqueue(self, data):
        new_node = Node(data)
        if self.tail is None:
            self.head = self.tail = new_node
            self.size += 1
            return        

        self.tail.next = new_node
        self.tail = self.tail.next
        self.size += 1
        return
    
    def dequeue(self):
        if self.isEmpty():
            return "Empty"
        
        if self.size == 1:
            temp = self.head.data 
            self.head = self.tail = None
            self.size -= 1
            return temp
        
        temp = self.head
        self.head = self.head.next
        self.size -= 1
        return temp.data
    
    def peek(self):
        if self.isEmpty():
            return "Empty"
        
        return self.tail.data
    
    def print_queue(self):
        temp = self.head
        while temp:
            print(temp.data, end='=>')
            temp = temp.next
        print()
    
```

```python
myQueue = LinkedQueue()

for _ in range(1, 20):
    myQueue.enqueue(np.random.randint(1,100))

print("Queue: ", end="")
myQueue.print_queue()
print("Peek: ", myQueue.peek())
print("Dequeue: ", myQueue.dequeue())
print("Queue after Dequeue: ", end="")
myQueue.print_queue()
print("isEmpty: ", myQueue.isEmpty())
print("Size: ", myQueue.length()) 
```
