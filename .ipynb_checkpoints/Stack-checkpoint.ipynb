```python
#STACK

class Stack:
    def __init__(self):
        self.stack = []
    
    def push(self, element):
        self.stack.append(element)
    
    def pop(self):
        if self.isEmpty():
            return "Empty"
        return self.stack.pop()

    def peek(self):
        if self.isEmpty():
            return "Empty"
        return self.stack[-1]
    
    def isEmpty(self):
        return len(self.stack) == 0
    
    def size(self):
        return len(self.stack)
```

```python
mystack = Stack()
```

```python
import numpy as np

for _ in range(1, 20):
    mystack.push(np.random.randint(1,100))

```

```python
print(mystack.peek())
```

```python
print("Stack: ", mystack.stack)
print("Pop: ", mystack.pop())
print("Stack after Pop: ", mystack.stack)
print("Peek: ", mystack.peek())
print("isEmpty: ", mystack.isEmpty())
print("Size: ", mystack.size()) 
```

```python
#STACK WITH LINK LIST

class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkStack:
    def __init__(self):
        self.head = None
        self.size = 0

    def isEmpty(self):
        return self.size == 0


    def push(self, value):
        new_node = Node(value)
        if self.head:
            new_node.next = self.head
        
        self.head = new_node
        self.size += 1

    def pop(self):
        if self.isEmpty():
            return "Empty"
        popped_node = self.head
        self.head = self.head.next
        self.size -=1

        return popped_node.value
    
    def peek(self):
        if self.isEmpty():
            return "Empty"
        
        return self.head.value

    def stackSize(self):
        return self.size

    def print_list(self):
        current_node = self.head
        while current_node:
            print(current_node.value, end="=>")
            current_node = current_node.next
        print()

        
```

```python
myStack = LinkStack()
for i in range(1, 20):
    myStack.push(np.random.randint(1,100))

print("LinkedList: ", end="")
myStack.print_list()
print("Peek: ", myStack.peek())
print("Pop: ", myStack.pop())
print("LinkedList after Pop: ", end="")
myStack.print_list()
print("isEmpty: ", myStack.isEmpty())
print("Size: ", myStack.stackSize()) 
```
