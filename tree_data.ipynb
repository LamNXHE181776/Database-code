```python
import numpy as np
```

```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    def __init__(self):
        self.root = None

    def insert(self, value):
        if self.root is None:
            self.root = TreeNode(value)
        else:
            self._insert_rec(self.root, value)

    def _insert_rec(self, node, value):
        if value < node.value:
            if node.left is None:
                node.left = TreeNode(value)
            else:
                self._insert_rec(node.left, value)
        else:
            if node.right is None:
                node.right = TreeNode(value)
            else:
                self._insert_rec(node.right, value)

    def inorder_traversal(self):
        return self._inorder_rec(self.root)
    
    def _inorder_rec(self, node):
        if node is None:
            return []
        return self._inorder_rec(node.left) + [node.value] + self._inorder_rec(node.right)
    
    def search(self, value):
        if self.root is None:
            return "Tree is Empty"
        else:
            return self.__search(self.root, value)

    def __search(self, node, value):
        if node is None:
            print("not found")
            return False
        if node.value == value:
            return True
        elif value < node.value:
            return self.__search(node.left, value)
        else:
            return self.__search(node.right, value)
        
    def lowest_value(self):
        if self.root is None:
            return "Tree is Empty"
        else:
            return self.__lowest_value(self.root)
        
    def __lowest_value(self, node):
        while node.left is not None:
            node = node.left

        return node
    
    def delete(self, value):
        if self.root is None:
            return "Tree is Empty"
        else:
            self.root = self.__delete(self.root, value)
        
        return True
        
    def __delete(self, node, value):
        if node is None:
            print("No value found")
            return None
        
        if node.value == value:
            if node.left is None:
                return node.right
            elif node.right is None:
                return node.left
            
            temp = self.__lowest_value(node.right)
            node.value = temp.value
            node.right = self.__delete(node.right, temp.value)

        elif value < node.value:
            node.left = self.__delete(node.left, value)
        else:
            node.right = self.__delete(node.right, value)
            
        return node
    
    def print_tree(self):
        # Helper function to print the tree structure with / \ lines
        lines = []
        self._print_rec(self.root, lines, "", "")
        for line in lines:
            print(line)
    def _print_rec(self, node, lines, prefix, children_prefix):
        if node is not None:
            lines.append(prefix + str(node.value))
            if node.left or node.right:
                if node.right:
                    self._print_rec(node.right, lines, children_prefix + "├── ", children_prefix + "│   ")
                else:
                    lines.append(children_prefix + "├── None")
                if node.left:
                    self._print_rec(node.left, lines, children_prefix + "└── ", children_prefix + "    ")
                else:
                    lines.append(children_prefix + "└── None")
```

```python
#insert values into the BST
bst = BinarySearchTree()
values = np.random.randint(0, 100, size=10)
for value in values:
    bst.insert(value)  
#perform inorder traversal
sorted_values = bst.inorder_traversal()
print("Inserted values:", values)
print("Inorder traversal (sorted values):", sorted_values)
```
